<h3 style="font-weight:bold; color:#ffffff;">Introduction to Ai Impact job Layoff Risk Analysis</h3>
<p >Artificial intelligence is now part of daily work in many jobs, from drafting reports to sorting invoices and answering support tickets. Companies adopt these tools to save time, but the same tools can take over tasks that people used to do. That raises a practical question for workers: which jobs are more likely to be cut?</p>
<p>The project compares risk across industries, roles and job levels, checks how task type and AI exposure relate to the label, and trains a classifier to predict risk from the other columns. The label comes with the dataset, so the results describe patterns in this data and do not forecast real layoffs.</p>
<p >This project uses Python libraries such as <span style="font-weight:bold;">Pandas</span>, <span style="font-weight:bold;">Numpy</span>, <span style="font-weight:bold;">Plotly</span>, <span style="font-weight:bold;">Matplotlib</span>, and <span style="font-weight:bold;">Seaborn</span> to analyze a stroke dataset.</p>

<h2 style="text-align:center;color:white;background-color:#0078d4;padding:10px;border-radius:10px;font-weight:600;">Data Preprocessing</h2>

Data preprocessing is the process of cleaning, transforming, and organizing raw data into a structured format so it is accurate and ready for data analysis. It includes <span style="font-weight:bold;font-style: italic;">Remove duplicate records, Handle missing values (NULLs), Correct data types, Remove irrelevant columns, Fix inconsistent values, Correct spelling errors, Remove or handle outliers, Standardize formats </span>

<h4>i. Handling Missing Values</h4>

```python
# Calculate total missing values
print("Total NULL values:", df.isnull().sum().sum())
```

 <span style="font-weight:bold;font-style: italic;">Info :</span> Used to find how many missing values are present in each column.

<h4>ii. Removing Duplicates</h4>

```python
# Check duplicate records
print("Duplicate records:", df.duplicated().sum())
```

<span style="font-weight:bold;font-style: italic;">Info :</span> counts and displays the total number of duplicate rows in a Pandas DataFrame. It is a common data-cleaning check used in Python to see if repeated entries exist before analyzing data

<h4>iii. Correct data types</h4>

 Assigning suitable data types to each column in a dataset according to the nature of its values. This ensures accurate calculations, efficient memory usage, proper data analysis, and better preparation for machine learning.

```python
# Check current data types
print(df.dtypes)
```

<span style="font-weight:bold;font-style: italic;">Info :</span> It display the data types of all columns in a DataFrame. It helps us verify whether each column has the correct data type before performing data analysis

```python
# Convert numerical columns
numeric_cols = [
    'Age',
    'Years_of_Experience',
    'Routine_Task_Percentage',
    'Creativity_Requirement',
    'Human_Interaction_Level',
    'Number_of_AI_Tools_Used',
    'AI_Usage_Hours_Per_Week',
    'Tasks_Automated_Percentage',
    'AI_Training_Hours'
]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

 
```

<span style="font-weight:bold;font-style: italic;">Info :</span>  convert selected columns into numerical data types so that you can perform mathematical calculations, statistical analysis
.

```python
# Convert categorical columns
categorical_cols = [
    'Education_Level',
    'Industry',
    'Job_Role',
    'Company_Size',
    'Job_Level',
    'AI_Adoption_Level',
    'Layoff_Risk'
]
for col in categorical_cols:
    df[col] = df[col].astype('category')

```

<span style="font-weight:bold;font-style: italic;">Info :</span>  convert selected columns into categorical data types. Categorical columns contain values that represent groups, classes, or categories rather than numerical measurements.

```python
print(df.dtypes)
```

<h4>iv. Remove irrelevant columns</h4>

Deleting unnecessary or unrelated columns from a dataset that do not provide useful information for the intended analysis. This helps simplify the dataset, reduce unnecessary data, and improve processing efficiency.

```python
print(df.columns.tolist())
```

<h4>v. Fix inconsistent values</h4>

It is a data cleaning process in which different representations of the same information are identified and standardized into a common format. It improves data quality, consistency, and accuracy, making the dataset suitable for further analysis

```python
# Standardize text values
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.title()
```

<span style="font-weight:bold;font-style: italic;">Info :</span>  converting text data into a consistent format by removing unnecessary spaces and making the capitalization uniform throughout the dataset.

```python
# Check unique values
for col in categorical_cols:
    print(col, df[col].unique())
```

<span style="font-weight:bold;font-style: italic;">Info :</span> Check every  column one by one and displays all the different values available in each column so that we can identify and correct inconsistent values.

<h4>vi. Standardize formats</h4>

converting data into a consistent and uniform format throughout a dataset so that all values follow the same structure or pattern.

```python
# Remove inconsistent capitalization
df['Education_Level'] = (
    df['Education_Level'].str.strip().str.title()
)
```

```python
#Convert all text into capitalize
df['AI_Adoption_Level'] = df['AI_Adoption_Level'].str.capitalize()
```

```python
df['Industry'] = df['Industry'].str.strip()
```

<h4>vii. Validate data accuracy</h4>

preprocessing technique used to verify whether the values in a dataset are correct, realistic, and logically consistent. It helps identify incorrect, invalid, or unrealistic values and improves the reliability and quality of the dataset before performing data analysis

```python
# Check Age
print(df[df['Age'] < 18])
```

```python
# Check negative experience
print(df[df['Years_of_Experience'] < 0])
```

```python
# Check percentage values
percentage_cols = [
    'Routine_Task_Percentage',
    'Creativity_Requirement',
    'Human_Interaction_Level',
    'Tasks_Automated_Percentage'
]

for col in percentage_cols:

    invalid = df[
        (df[col] < 0) |
        (df[col] > 100)
    ]

    print(col, "Invalid values:", len(invalid))
```

```python
# Check experience against age
invalid_experience = df[
    df['Years_of_Experience'] > (df['Age'] - 18)
]

print("Invalid experience records:", len(invalid_experience))
```
